```python
df = pd.read_csv('computer.csv')
df
```

```python
df['age'] = df['age'].str.replace('31...40','31-0')
df
```

```python
x = df.iloc[:,0:4]
y=df['buys_computer']
x
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(
x,
y,
test_size=0.2,
random_state=1)
```
